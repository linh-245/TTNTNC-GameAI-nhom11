```python
"""
Demo nhanh: Random vs Think1 vs AlphaBeta vs MCTS
Chay: python demo.py
"""

import sys
import os
sys.stdout.reconfigure(encoding='utf-8', errors='replace') if hasattr(sys.stdout, 'reconfigure') else None
try:
    ROOT = os.path.dirname(os.path.dirname(os.path.abspath(__file__)))
except NameError:               # chạy trong notebook thì không có __file__
    ROOT = os.getcwd()          # thư mục chứa demo.ipynb (gốc dự án)
sys.path.insert(0, ROOT)
```

```python
def demo_one_game():
    """Chạy một ván game và in ra bàn cờ."""
    print("\n" + "="*50)
    print("DEMO: AlphaBeta (depth=4) vs Random")
    print("="*50)
    reward, moves = one_conn_game(
        partial(alphabeta_agent, depth=4, use_eval=True),
        conn_random,
        render=True
    )
    if reward == 1:
        print("\n=> AlphaBeta (DO) THANG!")
    elif reward == -1:
        print("\n=> Random (VANG) THANG!")
    else:
        print("\n=> HOA!")
    print(f"Tổng số nước: {len(moves)}")
```

```python
def demo_stats():
    """Thống kê win rate qua 50 ván."""
    pairs = [
        ("Think1 vs Random",   conn_think1,  conn_random,  50),
        ("Think2 vs Think1",   conn_think2,  conn_think1,  50),
        ("AlphaBeta vs Think3", partial(alphabeta_agent, depth=4), conn_think3, 30),
        ("MCTS-100 vs Think2", partial(mcts_agent, n_simulations=100), conn_think2, 20),
    ]

    print("\n" + "="*60)
    print("THỐNG KÊ WIN RATE (mỗi cặp chia đều đi trước/đi sau)")
    print("="*60)
    print(f"{'Cặp đấu':<30} {'W':>5} {'L':>5} {'D':>5} {'Score':>7}")
    print("-"*60)

    for name, p1, p2, n in pairs:
        t0 = time.time()
        result = run_games(p1, p2, n_games=n, verbose=False)
        elapsed = time.time() - t0
        print(f"{name:<30} {result['wins']:>5} {result['losses']:>5} "
              f"{result['draws']:>5} {result['score']:>7.3f}  ({elapsed:.1f}s)")
```

```python
if __name__ == '__main__':
    demo_one_game()
    demo_stats()
```
